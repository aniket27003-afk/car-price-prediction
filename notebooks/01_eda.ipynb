# AutoValue AI — EDA
Covers PRD FR-03: prices by year, price vs km, distribution, by fuel, by transmission, correlations.

```python
import pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from src.data_loader import load_clean
df = load_clean('data/cardekho_used_cars.csv')
df.head(), df.shape
```

```python
df['selling_price'].hist(bins=40); plt.title('Price distribution (INR)'); plt.show()
```

```python
df.groupby('year')['selling_price'].median().plot(marker='o', title='Median price by manufacturing year'); plt.show()
```

```python
plt.scatter(df['km_driven'], df['selling_price'], alpha=0.2); plt.xlabel('km_driven'); plt.ylabel('selling_price'); plt.title('Price vs km driven'); plt.show()
```

```python
sns.boxplot(data=df, x='fuel', y='selling_price'); plt.ylim(0, 2000000); plt.title('Price by fuel'); plt.show()
sns.boxplot(data=df, x='transmission', y='selling_price'); plt.ylim(0, 2000000); plt.title('Price by transmission'); plt.show()
```

```python
df[['year','km_driven','car_age','selling_price']].corr()
sns.heatmap(df[['year','km_driven','car_age','selling_price']].corr(), annot=True); plt.show()
```
